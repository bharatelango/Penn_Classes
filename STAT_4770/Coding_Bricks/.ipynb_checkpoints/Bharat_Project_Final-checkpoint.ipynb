# Research Questions

**Y-variable:** the party that won each county in the 2016 presidential election (Democrat or Republican), a two-level categorical variable. The unit of analysis is the county (2,715 counties).

1. Do better-off counties vote Republican and worse-off counties vote Democratic?
2. Does income inequality (or food insecurity) separate the parties better than income level does?
3. Which economic hardship measures (unemployment, uninsured rate, high school graduation, housing problems) differ most between the parties?
4. Do these patterns reflect economics, or mainly urban vs. rural differences?

Finally, a **classification tree** will test which of these variables best predict a county's party.

# Hypothesis (Question 1)

### With more well-off counties wanting to keep their money, they would vote for less extreme taxation, which would side with the Republican party, while less well-off counties will want increased public services, therefore voting for the Democratic party.

We will test this against the data.

```python
# Read in all the libraries and data

import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# A slightly smaller default plot size so plots fit on a slide

plt.rcParams['figure.figsize'] = (6, 4)

# Colors used in every plot (blue = Democrat, red = Republican)

pal = {'Democrat': 'tab:blue', 'Republican': 'tab:red'}
```

```python
# Importing total data

df = pd.read_csv("project_4770_7770.csv")

print(df.shape)
print(df.columns)
df.head()
```

```python
# Only keeping important variables

df_imp = df[['Row.Label', 'Five-digit.FIPS.Code', 'State.FIPS.Code',
       'County.FIPS.Code', 'State.Abbreviation',
       'Election.Results.2016', 'Uninsured', 'High.School.Graduation',
       'Median.Household.Income', 'Severe.Housing.Problems', 'Unemployment',
       'Food.Insecurity.Quintile', 'Income.Inequality.Quartile']]

# Making a 0/1 column where 1 = Democrat county, so we can take averages later

df['Dem'] = (df['Election.Results.2016'] == 'Democrat').astype(int)

df_imp.head()
```

# The Y-variable: County Winner

- Republicans won **2,269 counties (83.6%)** and Democrats won **446 counties (16.4%)**.
- The classes are very unbalanced: a "model" that always guesses Republican would already be right about **84%** of the time. That is the baseline our tree has to beat.

```python
# Data summary of counties won for each party

print(df_imp['Election.Results.2016'].value_counts())

sns.countplot(x='Election.Results.2016', data=df_imp, palette = pal)
plt.title('Counties won by each party')
plt.show()
```

# Counties Are Not People

- The Democrat counties are only 16% of counties but contain about **55% of the total population**.
- Democrat counties are much bigger: the median county population is about **145,000** vs. about **27,000** for Republican counties.
- So the map is mostly Republican counties, but not mostly Republican people. Everything that follows is about *counties*.

```python
# Total population living in the counties each party won

pop_total = df.groupby('Election.Results.2016')['Population'].sum()

print(pop_total / pop_total.sum())

sns.barplot(x=pop_total.index, y=pop_total.values, palette=pal)
plt.title('Total Population by Winning Party')
plt.show()
```

# State Differences

- Many states are one-party at the county level. Every county in **AK, DC, HI, MA and VT** went Democrat, and every county in **OK, WV and WY** went Republican (KY and KS are above 97%).
- Some of the all-Democrat states have very few counties (DC has 1, AK has 2, HI has 4), so 100% is less surprising than it looks.
- Because the state is so closely tied to the vote, it is worth keeping in mind that the county patterns below mix together differences *within* and *between* states.

```python
# Checking top states for voting rep vs. dem

df_dum = pd.get_dummies(df_imp, columns=['Election.Results.2016'])

df_dem = df_dum.groupby('State.Abbreviation')['Election.Results.2016_Democrat'].mean().sort_values()

df_rep = df_dum.groupby('State.Abbreviation')['Election.Results.2016_Republican'].mean().sort_values()

print(df_dem.nlargest(5))
print(df_rep.nlargest(5))
```

```python
# Share of Republican-won counties in every state

plt.figure(figsize=(11, 4))
plt.xticks(rotation=90)
sns.barplot(df_rep, color="Red")
plt.ylabel('Proportion')
plt.xlabel('States')
plt.title('Proportion of Republican Counties by State') 
plt.show()
```

# The Predictors We Focus On

**Numeric:** median household income, unemployment, uninsured rate, high school graduation, severe housing problems.

**Ordinal:** food insecurity quintile and income inequality quartile. Higher numbers appear to mean *more* of the problem: counties in Q5 food insecurity and Q4 inequality have the lowest median incomes and the worst health (worth confirming in the data dictionary).

**First look (medians):** income, high school graduation and housing problems differ between the parties. Unemployment (4.5% vs. 4.4%) and uninsured rate (9.6% vs. 10.1%) are almost identical.

```python
# Median of each economic variable for Democrat vs. Republican counties

df_imp.groupby('Election.Results.2016')[['Median.Household.Income', 'Unemployment', 'Uninsured', 'High.School.Graduation', 'Severe.Housing.Problems']].median()
```

# Question 1: Do Richer Counties Vote Republican?

- The median income of Democrat counties (**about $54.5k**) is actually *higher* than for Republican counties (**about $48.6k**). That is the opposite of the hypothesis.
- But look at the spread: Democrat counties cover a much wider range of incomes. They are both some of the poorest and some of the richest counties.

```python
# How does the election results correlate to the other variables

sns.boxplot(x='Election.Results.2016', y='Median.Household.Income', palette = pal, data=df_imp)
plt.title('Median Household Income by Winning Party')
plt.show()
```

# Income Has a U-Shaped Relationship With the Vote

- About **28%** of the poorest-decile counties and **40%** of the richest-decile counties went Democrat.
- In the middle deciles (3 to 5) only about **6 to 8%** did.
- So the hypothesis is **not supported in a simple form**: both ends of the income scale lean Democrat more than the middle does. This is an association, not a cause.

```python
# Splitting counties into 10 equal groups by income (1 = poorest, 10 = richest)

df['Income_Decile'] = pd.qcut(df['Median.Household.Income'], 10, labels=False) + 1

# Proportion of Democrat counties in each income group

dem_by_decile = df.groupby('Income_Decile')['Dem'].mean()

print(dem_by_decile.round(3))

sns.barplot(x=dem_by_decile.index, y=dem_by_decile.values, color='Blue')
plt.xlabel('Income Decile (1 = poorest)')
plt.ylabel('Proportion of Democrat Counties')
plt.title('Democrat Share by Income Group')
plt.show()
```

# Question 2: Income Inequality

- Democrat share climbs steadily with inequality: **6.7%** in the least unequal quartile (Q1) up to **31.2%** in the most unequal (Q4).
- This is a cleaner pattern than income level, which was U-shaped.
- Q4 counties also have the lowest median income (about $42k), which fits with the poorest income decile leaning Democrat.

```python
# Proportion of Democrat counties in each income inequality quartile

dem_by_ineq = df.groupby('Income.Inequality.Quartile')['Dem'].mean()

print(dem_by_ineq.round(3))

sns.barplot(x=dem_by_ineq.index, y=dem_by_ineq.values, color='Orange')
plt.xlabel('Income Inequality Quartile')
plt.ylabel('Proportion of Democrat Counties')
plt.title('Democrat Share by Income Inequality')
plt.show()
```

# Food Insecurity

- Only the **most** food insecure counties (Q5) stand out: **28%** went Democrat, compared with **9 to 17%** in the other quintiles.
- The trend is not steady (Q4 is the lowest at 9%), so food insecurity looks less informative than inequality.

```python
# Proportion of Democrat counties in each food insecurity quintile

dem_by_food = df.groupby('Food.Insecurity.Quintile')['Dem'].mean()

print(dem_by_food.round(3))

sns.barplot(x=dem_by_food.index, y=dem_by_food.values, color='Green')
plt.xlabel('Food Insecurity Quintile')
plt.ylabel('Proportion of Democrat Counties')
plt.title('Democrat Share by Food Insecurity')
plt.show()
```

```python
sns.boxplot(x='Election.Results.2016', y='Severe.Housing.Problems', palette = pal, data=df_imp)
plt.title('Severe Housing Problems by Winning Party')
plt.show()
```

# High School Graduation

Republican counties have the higher graduation rate (median **90.2%** vs. **86.2%**). A moderate difference, with a lot of overlap.

```python
sns.boxplot(x='Election.Results.2016', y='High.School.Graduation', palette = pal, data=df_imp)
plt.title('High School Graduation by Winning Party')
plt.show()
```

# Uninsured Rate

Nearly the same in both groups (median **10.1%** Republican vs. **9.6%** Democrat), with heavy overlap.

```python
sns.boxplot(x='Election.Results.2016', y='Uninsured', palette = pal, data=df_imp)
plt.title('Uninsured by Winning Party')
plt.show()
```

# Unemployment

Essentially identical (median **4.4%** vs. **4.5%**). Whether a county has high or low unemployment tells us almost nothing about its party.

```python
sns.boxplot(x='Election.Results.2016', y='Unemployment', palette = pal, data=df_imp)
plt.title('Unemployment by Winning Party')
plt.show()
```

# Question 3 Takeaway

- **Housing problems** is the strongest separator, followed by **high school graduation**.
- **Unemployment and uninsured rate** hardly differ, so general "economic hardship" does not line up with party in a simple way.

# How the Predictors Relate to Each Other

Housing problems and unemployment are only weakly related (correlation about 0.24), but the Democrat (blue) counties sit toward the right, at higher housing problems, across the whole range of unemployment.

```python
# Relationship between a variable of interest

sns.scatterplot(data=df, x='Severe.Housing.Problems', y='Unemployment', hue='Election.Results.2016', palette=pal, alpha = 0.5)
plt.title('Unemployment vs Housing Issues')
plt.show()
```

# Housing Problems vs. Income

```python
# Income vs. housing problems

sns.scatterplot(data=df, x='Median.Household.Income', y='Severe.Housing.Problems', hue='Election.Results.2016', palette=pal, alpha = 0.5)
plt.title('Housing Issues vs Income')
plt.show()
```

# Correlations Between the Economic Variables

The strongest relationships are income with unemployment (-0.44), income with the uninsured rate (-0.36), and high school graduation with housing problems (-0.33). All other pairs are weak (0.25 or less in size), so the variables are not simply copies of one another.

# Question 4: Is It Really Urban vs. Rural?

The biggest gap yet: the median Republican county is about **59% rural**, while the median Democrat county is only about **17% rural**.

Percent rural was not in our economic variable set, but it separates the parties more than any economic variable we looked at.

```python
# Percent of the county that is rural, by winning party

sns.boxplot(x='Election.Results.2016', y='Percent.Rural', palette = pal, data=df)
plt.title('Percent Rural by Winning Party')
plt.show()
```

# Income vs. How Rural a County Is

Higher income counties tend to be *less* rural (correlation about -0.41). The very high income counties (top left, many in blue) are mostly urban, so the right side of the income U-shape may partly reflect urban vs. rural differences. These plots cannot separate the two, so we should not over-interpret income on its own.

```python
# Income vs. how rural a county is

sns.scatterplot(data=df, x='Percent.Rural', y='Median.Household.Income', hue='Election.Results.2016', palette=pal, alpha = 0.5)
plt.title('Income vs Percent Rural')
plt.show()
```

# Classification Tree: Predicting the Party

- **Y:** winning party (kept as the text labels Democrat / Republican).
- **X:** the five numeric economic variables plus dummy columns for food insecurity and income inequality (14 columns total).
- **Setup:** 80/20 train/test split (2,172 training counties, 543 test counties), maximum of **10 leaves** to keep the tree readable.
- **Colors:** orange = Democrat, blue = Republican; darker means a purer group. Each box shows the *proportion* of counties in it.

```python
# Tree full code for classification
from sklearn.model_selection import train_test_split
from sklearn import tree
import sklearn

Xcts = df_imp[['Uninsured', 'High.School.Graduation', 'Median.Household.Income',
       'Severe.Housing.Problems', 'Unemployment']]

Xcats = df_imp[['Food.Insecurity.Quintile', 'Income.Inequality.Quartile']]

dummies = pd.get_dummies(Xcats)

X = Xcts.join(dummies)

y = df_imp['Election.Results.2016']

# Train split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state=2006)

# Tree training

reg_tree = tree.DecisionTreeClassifier(max_leaf_nodes=10, random_state=2006)
reg_tree.fit(X_train, y_train)

fig = plt.figure(num=None, figsize=(12, 9))
tree.plot_tree(reg_tree, filled=True, feature_names=list(X.columns), class_names=list(reg_tree.classes_), proportion=True, impurity=False)
plt.show()
```

# Reading the Tree

- **First split: severe housing problems <= 17.5%.** About **82%** of counties go left, and that group is already **91% Republican**.
- **Largest leaf:** housing <= 14.9%. This is **60% of all counties, and 96% are Republican**. These are the easy ones.
- **Right branch** (housing > 17.5%, 18% of counties) is **54% Democrat**. If median income is also above about $56.5k, the leaf is **77% Democrat**.
- Lower-income counties with high housing problems are mixed, and the tree uses income inequality (Q4) and the uninsured rate to sort them. Leaves of 1 to 3% of counties are close to coin flips and should not be over-read.

# How Well Does the Tree Predict?

- Accuracy on the 543 held-out test counties is **87.8%**, compared with **84.3%** from always guessing Republican. A real but modest improvement.
- The tree catches **45 of the 85** Democrat test counties (53%); when it predicts Democrat, it is right 45 out of 71 times (63%).
- It is much better at Republican counties (432 of 458 correct).

```python
# How well does the tree predict counties it has never seen?

print('Tree accuracy on test counties:', reg_tree.score(X_test, y_test))
print('Accuracy from always guessing Republican:', (y_test == 'Republican').mean())

# Actual vs. predicted party

pd.crosstab(y_test, reg_tree.predict(X_test), rownames=['Actual'], colnames=['Predicted'])
```

# Variable Importance

- **Severe housing problems (0.63)** and **median income (0.19)** account for about 80% of the tree's importance.
- Uninsured rate, income inequality, unemployment and food insecurity each contribute little (0.03 to 0.06), and high school graduation is never used.
- The importance ranking agrees with Question 3: housing problems is the strongest economic separator.

```python
# Rank the variables by importance

imp = pd.Series(reg_tree.feature_importances_, index=X.columns).sort_values(ascending=False)

# Putting the dummy columns back together into one bar per original variable

imp_grouped = imp.groupby(lambda c: c.split('_Q')[0]).sum().sort_values()

print(imp_grouped.sort_values(ascending=False).round(3))

imp_grouped.plot(kind='barh', figsize=(7, 4.5), color='Green')
plt.title('Variable Importance')
plt.show()
```

```python
# Same ranking, but with every dummy column shown separately

imp.sort_values().plot(kind='barh', figsize=(7, 4.5), color='tab:blue')
plt.title('Variable Importance (all columns)')
plt.show()
```

# Conclusions

1. **Hypothesis not supported in a simple form.** Democrat counties have a higher median income ($54.5k vs. $48.6k), and the Democrat share is highest at *both* ends of the income scale (U-shape).
2. **Inequality** shows a steady climb (6.7% to 31.2% Democrat across quartiles). Food insecurity only stands out in its top quintile.
3. **Housing problems** (18.3% vs. 13.3%) and **high school graduation** separate the parties most; **unemployment and uninsured rate barely differ**.
4. **Urban vs. rural** (59% vs. 17% rural) is a bigger gap than any economic variable, so part of the economic story may really be an urbanization story.
5. A 10-leaf **tree** reaches **87.8%** test accuracy (baseline 84.3%), driven by housing problems and income.

# Caveats and Next Steps

- These are **county-level, observational associations** from one election. They are not causal, and they say nothing about how individual voters behave.
- The Democrat counties are few but populous, so "counties won" is not the same as "votes won".
- The tree used one train/test split and a fixed leaf limit, and the importance scores describe *this* tree. Correlated variables can share or hide credit.
- **Next step:** add Percent Rural and Population to the tree and see whether they take over from the economic variables.
